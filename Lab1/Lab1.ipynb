````python
a = 15
b = 4
c = a + b
print("Сумма:", c)
d = a - b
print("Разность:", d)
e = a * b
print("Произведение:", e)
f = a / b
print("Частное:", f)
g = a // b
print("Целое частное:", g)
h = a % b
print("Остаток:", h)
i = a ** b
print("Возведение в степень:", i)
````
